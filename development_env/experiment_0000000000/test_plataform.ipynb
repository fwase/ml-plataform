```python
class ModelRepoSingleton():
    _instance = None

    def __init__(self):
        if self._instance is not None:
            print("Getting from cache")
            return 
        print("Init class")
        self.attribute_a = self.load_variable_a()

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

    def load_variable_a(self):
        print("Getting variable a")
        return 1.0
```
